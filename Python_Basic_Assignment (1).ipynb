7. What is the difference between == and is operators in Python?
   1. == checks if values of two variables are equal.

  2. 'is' checks if two variables refer to the same object in memory.



```python
a = [1, 2, 3]
b = [1, 2, 3]
print(a == b)  # True (same values)
print(a is b)  # False (different memory locations)
```

10. What is the difference between implicit and explicit type casting?

  1.Implicit Type Casting: Python automatically converts a smaller data type to
    a larger one (e.g., int to float).

  2.Explicit Type Casting: The programmer manually converts a data type using
    functions like int(), float(), str().



```python
# Example Implicit and Explicit Type casting
# 1. Implicit Type Casting
x = 10
y = 2.5
z = x + y
print(type(z))

#2. Explicit Type Casting
x = "123"
y = int(x)
print(type(x))
print(type(y))
```

11. What is the purpose of conditional statements in Python?

  Conditional statements allow decision-making in programs by executing different code blocks based on conditions. Examples include if, elif, and else.



```python
# Example Conditional statement
x = 10
if x > 8:
    print("x is greater than 8")
elif x == 8:
    print("x is equal to 8")
else:
    print("x is less than 8")
```

13. What is the difference between for and while loops?

  1.for loop is used when the number of iterations is known beforehand.

  2.while loop is used when the number of iterations is unknown and depends on  
    a condition.


```python
# Example While and For Loop
for i in range(5):  # Runs 5 times
    print(i)

x = 0
while x < 5:  # Runs until x reaches 5
    print(x)
    x += 1
```

14. Describe a scenario where a while loop is more suitable than a for loop.

    A while loop is more suitable when the number of iterations is unknown and depends on user input or a condition.

```python
# Example
user_input = ""
while user_input != "exit":
    user_input = input("Enter a command (type 'exit' to quit): ")
    print("You entered:", user_input)
```

```python
# 14Q Write a program to reverse a string using a while loop.
input_string = input("Enter a string: ")
reversed_string = ""
index = len(input_string) - 1
while index >= 0:
    reversed_string += input_string[index]
    index -= 1
print("Reversed string:", reversed_string)
```

```python
#15.Write a Python program to calculate the factorial of a number provided by the user using a while loop.
num = int(input("Enter a number: "))
factorial = 1
current = 1
while current <= num:
    factorial *= current
    current += 1
print(f"The factorial of {num} is {factorial}")
```
